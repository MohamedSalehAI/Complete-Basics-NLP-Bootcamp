```python
import requests
from bs4 import BeautifulSoup

def GetPage(Link, FileName):
    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/138.0.0.0 Safari/537.36"
        )
    }

    page = requests.get(Link, headers=headers, timeout=10)
    page.raise_for_status()

    soup = BeautifulSoup(page.text, "html.parser")

    print("Status Code:", page.status_code)

    paragraphs = soup.find_all("p")
    print(f"Number of paragraphs: {len(paragraphs)}")

    title = soup.find(id="firstHeading")
    if title:
        print("Page title:", title.get_text())

    with open(FileName, "w", encoding="utf-8") as f:
        for p in paragraphs:
            f.write(p.get_text())
            f.write("\n")
GetPage(
    "https://en.wikipedia.org/wiki/Python_(programming_language)",
    "python.txt"
)
```

```python
Link='https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population'
headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/138.0.0.0 Safari/537.36"
        )
    }

page = requests.get(Link, headers=headers, timeout=10)
page.raise_for_status()

soup = BeautifulSoup(page.text, "html.parser")
```

```python
def process_num(num):
    return float(re.sub(r'[^\w\s.]','',num))
```

```python
ranks = []
rates = []
countries = []

for table in tables:
    rows = table.find_all('tr')


    for row in rows:
        cells = row.find_all('td')


        if len(cells) > 2: # Ensure there are at least three cells for rank, country, and rate
            rank_text = cells[0].text.strip()
            country_text = cells[1].text.strip()
            rate_text = cells[2].text.strip()

            try:
                processed_rate = process_num(rate_text)
                # Only append if all values are successfully processed
                ranks.append(rank_text)
                countries.append(country_text)
                rates.append(processed_rate)
            except ValueError:
                # If the 'rate' text cannot be converted to a float, skip this entry.
                # This handles cases where cells[2] contains non-numeric data like dates ('18 March').
                print(f"Skipping row with non-numeric rate (index 2): '{rate_text}'")
```

```python
print(ranks)
print(countries)
print(rates)
```

```python
data = pd.DataFrame(ranks, index= countries ,columns = ['Rank'])
data['DALY rate'] = rates
data.head(10)
```
