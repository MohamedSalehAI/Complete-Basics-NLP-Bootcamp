```python
synonyms = []
antonyms = []

for syn in wordnet.synsets("good"):
    for l in syn.lemmas():
        synonyms.append(l.name())
        if l.antonyms():
            antonyms.append(l.antonyms()[0].name())

print(set(synonyms))
print('------------------------------------')
print(set(antonyms))
```

```python
from nltk.tokenize import word_tokenize


text = 'I need to go right now to my friend Hasan love'

for word in word_tokenize(text) :
    synonyms = []
    antonyms = []


    for syn in wordnet.synsets(word):
        for l in syn.lemmas():
            synonyms.append(l.name())
            if l.antonyms():
                antonyms.append(l.antonyms()[0].name())

    print(f'word : {word}')
    if len(synonyms) >1 :print(f'\nsynonyms : {set(synonyms)}')
    if len(antonyms) >1: print(f'\nantonyms : {set(antonyms)}\n')
    print('===========================================')
```

```python
# word1 = 'green'
# word2 = 'blue'

# word1 = 'moon'
# word2 = 'sun'

word1 = 'water'
word2 = 'air'

# word1 = 'france'
# word2 = 'paris'

# word1 = 'lion'
# word2 = 'shoes'

wd1=wordnet.synsets(word1)[0]
wd2=wordnet.synsets(word2)[0]
print(wd1)
print(wd2)
wd3=wordnet.synset(wd1.name())
wd4=wordnet.synset(wd2.name())
print(wd3)
print(wd4)
print(wd3.wup_similarity(wd4))
w1 = wordnet.synset(wordnet.synsets(word1)[0].name())
w2 = wordnet.synset(wordnet.synsets(word2)[0].name())

print(f'Syn ({wordnet.synsets(word1)[0].name()}) and ({wordnet.synsets(word2)[0].name()}) with similarity {wd1.wup_similarity(wd2)}')
```

```python
text = 'I went to the school that day to talk to the teacher'

for word1 in word_tokenize(text) :
    for word2 in word_tokenize(text) :
        if len(wordnet.synsets(word1))>0 and len(wordnet.synsets(word2))>0  :
            w1 = wordnet.synset(wordnet.synsets(word1)[0].name())
            w2 = wordnet.synset(wordnet.synsets(word2)[0].name())
            print(f'Syn ({wordnet.synsets(word1)[0].name()}) and ({wordnet.synsets(word2)[0].name()}) with similarity {w1.wup_similarity(w2)}')
            print('----------------------------------------')
```
