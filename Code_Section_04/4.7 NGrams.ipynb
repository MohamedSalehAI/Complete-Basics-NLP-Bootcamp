```python
import nltk as nl
from nltk.util import ngrams

string = "moses supposes his toeses are roses but moses supposes erroneously"
bigrams1 = nl.bigrams(string)
bigrams2 = ngrams(string,2)
for i in bigrams1:
    print(i)
print('========================')
for i in bigrams2:
    print(i)
```

```python
bigrams1 = nl.bigrams(string.split())
bigrams2 = ngrams(string.split(),2)
for i in bigrams1:
    print(i)
print('========================')
for i in bigrams2:
    print(i)
```

```python
trigrams1 = nl.trigrams(string.split())
trigrams2 = ngrams(string.split(),3)
for i in trigrams1:
    print(i)
print('========================')
for i in trigrams2:
    print(i)
```

```python
from itertools import tee, islice

words = ["I", "love", "machine", "learning"]

a, b = tee(words)
#islice(iterable,start,stop,step) or islice(iterable,stop)
# print(list(a))
# print(list(b))

b = islice(b, 1, None)

print(list(zip(a, b)))
```
