# Distance Metrics

## 1. Euclidean Distance

Given two vectors

$$
\mathbf{a} = (a_1, a_2, \ldots, a_n)
$$

and

$$
\mathbf{b} = (b_1, b_2, \ldots, b_n),
$$

their Euclidean distance is

$$
d_{\text{euclidean}}(\mathbf{a}, \mathbf{b})
=
\|\mathbf{a}-\mathbf{b}\|_2
=
\sqrt{\sum_{i=1}^{n}(a_i-b_i)^2}
$$

---

## 2. Dot Product

The dot product of two vectors is

$$
\mathbf{a}\cdot\mathbf{b}
=
\sum_{i=1}^{n}a_i b_i
$$

---

## 3. Vector Norm (Magnitude)

The Euclidean norm of a vector is

$$
\|\mathbf{a}\|
=
\sqrt{\sum_{i=1}^{n}a_i^2}
$$

Similarly,

$$
\|\mathbf{b}\|
=
\sqrt{\sum_{i=1}^{n}b_i^2}
$$

---

## 4. Cosine Similarity

Cosine similarity measures the angle between two vectors:

$$
\operatorname{cos\_sim}(\mathbf{a},\mathbf{b})
=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{a}\|\|\mathbf{b}\|}
$$

or equivalently,

$$
\operatorname{cos\_sim}(\mathbf{a},\mathbf{b})
=
\frac{\sum_{i=1}^{n}a_i b_i}
{\sqrt{\sum_{i=1}^{n}a_i^2}
\sqrt{\sum_{i=1}^{n}b_i^2}}
$$

---

## 5. Cosine Distance

Cosine distance is defined as

$$
d_{\text{cosine}}(\mathbf{a},\mathbf{b})
=
1-
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{a}\|\|\mathbf{b}\|}
$$

or

$$
d_{\text{cosine}}(\mathbf{a},\mathbf{b})
=
1-
\frac{\sum_{i=1}^{n}a_i b_i}
{\sqrt{\sum_{i=1}^{n}a_i^2}
\sqrt{\sum_{i=1}^{n}b_i^2}}
$$

---

## Relation

$$
\boxed{
d_{\text{cosine}}
=
1-\operatorname{cos\_sim}
}
$$

# Pairwise Distances

The function

```python
from sklearn.metrics.pairwise import pairwise_distances
```

computes the distance between **every pair of vectors** in two datasets.

---

## Input

Suppose we have

$$
X =
\begin{bmatrix}
x_1 \\
x_2 \\
\vdots \\
x_m
\end{bmatrix}
\in \mathbb{R}^{m \times d}
$$

and

$$
Y =
\begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_n
\end{bmatrix}
\in \mathbb{R}^{n \times d}
$$

where:

- \(m\): number of samples in \(X\)
- \(n\): number of samples in \(Y\)
- \(d\): number of features

---

## Output

The function returns a **distance matrix**

$$
D \in \mathbb{R}^{m \times n}
$$

where each element is

$$
D_{ij}=d(x_i,y_j)
$$

That is, the distance between the \(i\)-th sample of \(X\) and the \(j\)-th sample of \(Y\).

---

## Euclidean Distance

```python
pairwise_distances(X, Y, metric="euclidean")
```

$$
D_{ij}
=
\sqrt{\sum_{k=1}^{d}(x_{ik}-y_{jk})^2}
$$

---

## Manhattan Distance

```python
pairwise_distances(X, Y, metric="manhattan")
```

$$
D_{ij}
=
\sum_{k=1}^{d}|x_{ik}-y_{jk}|
$$

---

## Cosine Distance

```python
pairwise_distances(X, Y, metric="cosine")
```

$$
D_{ij}
=
1-
\frac{x_i\cdot y_j}
{\|x_i\|\|y_j\|}
$$

---

## Minkowski Distance

```python
pairwise_distances(X, Y, metric="minkowski", p=p)
```

$$
D_{ij}
=
\left(
\sum_{k=1}^{d}
|x_{ik}-y_{jk}|^p
\right)^{\frac{1}{p}}
$$

Special cases:

- \(p=1\): Manhattan distance
- \(p=2\): Euclidean distance

---

## General Equation

For any metric,

$$
\boxed{
D_{ij}=d(x_i,y_j)
}
$$

where \(d(\cdot,\cdot)\) is the selected distance function.

---

## Example

If

$$
X=
\begin{bmatrix}
1 & 2\\
3 & 4
\end{bmatrix},
\qquad
Y=
\begin{bmatrix}
5 & 6\\
7 & 8
\end{bmatrix}
$$

then

```python
pairwise_distances(X, Y, metric="euclidean")
```

returns

$$
D=
\begin{bmatrix}
5.657 & 8.485\\
2.828 & 5.657
\end{bmatrix}
$$

where

$$
D_{11}=d(x_1,y_1),\quad
D_{12}=d(x_1,y_2),\quad
D_{21}=d(x_2,y_1),\quad
D_{22}=d(x_2,y_2).
$$
