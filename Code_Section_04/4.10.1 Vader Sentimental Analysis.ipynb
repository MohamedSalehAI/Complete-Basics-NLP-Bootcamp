```python
import nltk
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
SIA = SentimentIntensityAnalyzer()
```

```python
import numpy as np
```

```python
a = 'This was a good movie.'


eval=dict((b,list(dict(sorted(SIA.polarity_scores(b).items(),key=lambda x:x[1])[::-1]))[0]) for b in a.split())
eval2=list(dict(sorted(SIA.polarity_scores(a).items(),key=lambda x:x[1])[::-1]))[0]
print(eval2)
print(eval)
SIA.polarity_scores(a)
```
