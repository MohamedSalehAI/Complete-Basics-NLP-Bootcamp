```python
import pandas as pd
npr = pd.read_csv('npr.csv')
npr.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import NMF

documents = [
    "I love football and basketball",
    "The team scored many goals",
    "Python is used in machine learning",
    "Deep learning uses neural networks"
]

vectorizer = TfidfVectorizer(stop_words="english")
X = vectorizer.fit_transform(documents)

nmf = NMF(n_components=2, random_state=42)

W = nmf.fit_transform(X)
H = nmf.components_

words = vectorizer.get_feature_names_out()

for i, topic in enumerate(H):
    print(f"Topic {i+1}:")
    print([words[j] for j in topic.argsort()[-5:][::-1]])
for idx, doc in enumerate(W):
    print("propability is :",doc)
    print(f"Document {idx+1} belongs to topic {doc.argmax()+1}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm = tfidf.fit_transform(npr['Article'])
```

```python
from sklearn.decomposition import NMF
nmf_model = NMF(n_components=7,random_state=42)
nmf_model.fit_transform(dtm)
```
