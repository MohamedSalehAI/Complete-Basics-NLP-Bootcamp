```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_df=0.95, min_df=2, stop_words='english')
dtm = cv.fit_transform(npr['Article'])
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
documents = [
    "football team goal match coach",
    "basketball player score game",
    "government election president vote",
    "minister parliament politics law",
    "python machine learning neural network",
    "deep learning tensorflow pytorch",
    "football stadium striker coach",
    "AI transformer language model"
]
vectorizer = CountVectorizer(stop_words="english")

X = vectorizer.fit_transform(documents)
lda = LatentDirichletAllocation(
    n_components=3,
    random_state=42
)
results=lda.fit_transform(X)
print(lda.components_.shape)
# print(lda.components_)
feature_names = vectorizer.get_feature_names_out()

for topic_idx, topic in enumerate(lda.components_):
    print(f"Topic {topic_idx}")

    top_words = topic.argsort()[::-1][:5]

    print(feature_names[top_words])
for i,prob in enumerate(results):
    print(i,">>>>",prob)
most=results.argmax(axis=1)
df=pd.DataFrame(most,columns=['topic'])
print(df)
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

for index, topic in enumerate(LDA.components_):
    print(f"THE TOP 15 WORDS FOR TOPIC #{index+1}")

    words = [cv.get_feature_names_out()[i] for i in topic.argsort()[-15:]]

    doc = nlp(" ".join(words))

    print(", ".join(
        token.lemma_.lower()
        for token in doc
        if token.is_alpha
    ))

    print()
```
