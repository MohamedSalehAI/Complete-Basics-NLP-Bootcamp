| String     | `isspace()` |
| ---------- | ----------- |
| `" "`      | True        |
| `"     "`  | True        |
| `"\t"`     | True        |
| `"\n"`     | True        |
| `" \t\n "` | True        |
| `"Hello"`  | False       |
| `"Hello "` | False       |
| `""`       | False       |


```python
blanks = []

for idx,label,reviw in df.itertuples():
    if isinstance(label,str):
        if reviw.isspace() and reviw.strip()=="":
            blanks.append(idx)

print(len(blanks), 'blanks: ', blanks)
```

```python
df = df[df["review"].apply(
    lambda x: not (isinstance(x, str) and x.strip() == "")
)]
len(df)
```

```python
df['label'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
X = df['review']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
y_train.value_counts() , y_test.value_counts()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

text_clf_nb = Pipeline([('tfidf', TfidfVectorizer()),
                     ('clf', MultinomialNB()),])
text_clf_nb_cv = Pipeline([('cv', CountVectorizer()),
                     ('clf', MultinomialNB()),])

text_clf_lsvc = Pipeline([('tfidf', TfidfVectorizer()),
                     ('clf', LinearSVC()),])
text_clf_lsvc_cv = Pipeline([('cv', CountVectorizer()),
                     ('clf', LinearSVC()),])

text_clf_nb.fit(X_train, y_train)
# text_clf_nb_cv.fit(X_train, y_train)
# text_clf_lsvc.fit(X_train, y_train)
# text_clf_lsvc_cv.fit(X_train, y_train)
```

```python
predictions = text_clf_nb.predict(X_test)
```

```python
from sklearn import metrics
print(metrics.confusion_matrix(y_test,predictions))
```

```python
print(metrics.classification_report(y_test,predictions))
```

```python
print(metrics.accuracy_score(y_test,predictions))
```

```python
text_clf_lsvc.fit(X_train, y_train)
```

```python
predictions = text_clf_lsvc.predict(X_test)
```

```python
print(metrics.confusion_matrix(y_test,predictions))
```

```python
print(metrics.classification_report(y_test,predictions))
```

```python
print(metrics.accuracy_score(y_test,predictions))
```

```python
text_clf_lsvc_cv.fit(X_train, y_train)
pred=text_clf_lsvc_cv.predict(X_test)
print(metrics.confusion_matrix(y_test,pred))
print(metrics.classification_report(y_test,pred))
print(metrics.accuracy_score(y_test,pred))
#
```
