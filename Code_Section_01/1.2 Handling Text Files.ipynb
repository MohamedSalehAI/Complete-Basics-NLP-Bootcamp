| Mode   | Meaning                 | File must exist? | Creates file? | Erases existing content? |
| ------ | ----------------------- | ---------------- | ------------- | ------------------------ |
| `"r"`  | Read                    | ✅ Yes            | ❌ No          | ❌ No                     |
| `"w"`  | Write                   | ❌ No             | ✅ Yes         | ✅ Yes                    |
| `"a"`  | Append                  | ❌ No             | ✅ Yes         | ❌ No                     |
| `"x"`  | Create only             | ❌ Must not exist | ✅ Yes         | ❌ No                     |
| `"r+"` | Read and write          | ✅ Yes            | ❌ No          | ❌ No                     |
| `"w+"` | Read and write          | ❌ No             | ✅ Yes         | ✅ Yes                    |
| `"a+"` | Read and append         | ❌ No             | ✅ Yes         | ❌ No                     |
| `"x+"` | Create, read, and write | ❌ Must not exist | ✅ Yes         | ❌ No                     |
