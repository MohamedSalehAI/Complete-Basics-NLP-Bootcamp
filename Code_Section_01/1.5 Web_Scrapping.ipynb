#Timeout

```python
import requests

try:
    response = requests.get(
        "https://httpbin.org/delay/5",
        timeout=200
    )
except requests.exceptions.Timeout:
    print("Request timed out")
```

```python
print(response.status_code)   # HTTP status code
print(response.headers)       # Response headers
print(response.text)          # Text content (HTML, JSON, etc.)
print(response.content)       # Raw bytes (images, PDFs, etc.)
print(response.url)           # Final URL after redirects
print(response.encoding)      # Character encoding
print(response.cookies)       # Cookies sent by the server
print(response.elapsed)       # Time taken for the request
print(response.ok)            # True if status code < 400
```

#Download an Image

```python
import requests
import matplotlib.pyplot as plt

url = "https://images.unsplash.com/photo-1526779259212-939e64788e3c?fm=jpg&q=60&w=3000&auto=format&fit=crop&ixlib=rb-4.1.0&ixid=M3wxMjA3fDB8MHxzZWFyY2h8M3x8ZnJlZSUyMGltYWdlc3xlbnwwfHwwfHx8MA%3D%3D"

r = requests.get(url)

if r.status_code == 200 and r.headers["Content-Type"].startswith("image/"):

    with open("image.jpg", "wb") as f:
        f.write(r.content)

    img = plt.imread(open("image.jpg", "rb"))

    plt.imshow(img)
    plt.axis("off")
    plt.show()

else:
    print("Download failed")
```

#Module 4 — Pagination

```python
import requests
from bs4 import BeautifulSoup

for page in range(1,6):

    url=f"https://example.com/products?page={page}"

    print(url)
```

```python
import requests
from bs4 import BeautifulSoup

all_products=[]

for page in range(1,6):

    url=f"https://example.com/products?page={page}"

    response=requests.get(url)

    soup=BeautifulSoup(response.text,"html.parser")

    products=soup.select(".product")

    for product in products:

        all_products.append(product.text.strip())

print(all_products)
```

```python
import csv
page=1
with open("products.csv","w",newline="") as file:

    writer=csv.writer(file)

    writer.writerow(["Product"])


    while True:

             url=f"https://example.com?page={page}"

             response=requests.get(url)

             soup=BeautifulSoup(response.text,"html.parser")

             products=soup.select(".product")

             if len(products)==0:

                 break

             print(page)

             page+=1

             for product in products:

               writer.writerow([product.text.strip()])
```

#Project for fetching Github

```python
import requests
import pandas as pd

BASE_URL = "https://api.github.com/search/repositories"

keyword = input("Enter repository name: ")

params = {
    "q": keyword,
    "sort": "stars",
    "order": "desc",
    "page": 1,
    "per_page": 10
}

response = requests.get(
    BASE_URL,
    params=params,
    timeout=10
)

response.raise_for_status()

data = response.json()
print("response",data)
repositories = data["items"]

rows = []

for repo in repositories:

    print("-" * 50)

    print("Name:", repo["name"])
    print("Owner:", repo["owner"]["login"])
    print("Stars:", repo["stargazers_count"])
    print("Language:", repo["language"])
    print("URL:", repo["html_url"])
    print()

    rows.append({
        "Name": repo["name"],
        "Owner": repo["owner"]["login"],
        "Stars": repo["stargazers_count"],
        "Language": repo["language"],
        "URL": repo["html_url"]
    })

df = pd.DataFrame(rows)

df.to_csv("repositories.csv", index=False)

print("\nSaved to repositories.csv")
```

```python
import requests
import pandas as pd

BASE_URL = "https://data.usajobs.gov/api/search"

keyword = input("Enter job title: ")

headers = {
    "Host": "data.usajobs.gov",
    "User-Agent": "your_email@example.com",
    "Authorization-Key": "YOUR_API_KEY"
}

params = {
    "Keyword": keyword,
    "ResultsPerPage": 10
}

response = requests.get(
    BASE_URL,
    headers=headers,
    params=params,
    timeout=10
)

# response.raise_for_status()

data = response.json()

jobs = data["SearchResult"]["SearchResultItems"]

rows = []

for job in jobs:

    info = job["MatchedObjectDescriptor"]

    print("-" * 50)

    print("Title:", info["PositionTitle"])
    print("Organization:", info["OrganizationName"])
    print("Location:", info["PositionLocation"][0]["LocationName"])
    print("Apply:", info["PositionURI"])

    rows.append({
        "Title": info["PositionTitle"],
        "Organization": info["OrganizationName"],
        "Location": info["PositionLocation"][0]["LocationName"],
        "Apply URL": info["PositionURI"]
    })

df = pd.DataFrame(rows)

df.to_csv("jobs.csv", index=False)

print("Saved to jobs.csv")
```
