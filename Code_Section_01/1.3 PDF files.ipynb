#Read a Pdf

```python
import os

path="/content/US_Declaration.pdf"

if not os.path.exists(path):
    raise FileNotFoundError(f"The file was not found at the specified path: {path}")

if os.path.getsize(path) == 0:
    raise ValueError(f"The file at '{path}' is empty. Please check the file content.")

f = open(path,'rb')
pdf_reader = PyPDF2.PdfReader(f)
```

```python
print("num of pages is :")
len(pdf_reader.pages)
```

```python
pdf_reader.pages
```

```python
page_one = pdf_reader.pages[0]
page_one_text = page_one.extract_text()
print(page_one_text)
```

```python
page_two=pdf_reader.pages[1]
text=page_two.extract_text()
list=text.split()
print(list)
```

```python
f = open('US_Declaration.pdf','rb'  )

# List of every page's text.
# The index will correspond to the page number.
pdf_text = [0]  # zero is a placehoder to make page 1 = index 1

pdf_reader = PyPDF2.PdfReader(f)

for p in range(len(pdf_reader.pages)):

    page = pdf_reader.pages[p]

    pdf_text.append(page.extract_text())

f.close()

print(pdf_text)
print(len(pdf_text))
```

#Merge a PDF

```python
from PyPDF2 import PdfMerger, PdfReader,  PdfWriter

filename1 = '/content/drive/MyDrive/Colab Notebooks/Basics_of_NLP/Code_Section_01/F1.pdf'
filename2 = '/content/drive/MyDrive/Colab Notebooks/Basics_of_NLP/Code_Section_01/F2.pdf'

merger = PdfMerger()

merger.append(PdfReader(open(filename1, 'rb')))
merger.append(PdfReader(open(filename2, 'rb')))

merger.write("merged file 1.pdf")
```

```python
merger=open('merged file 1.pdf','rb')
pdf_reader=PyPDF2.PdfReader(merger)
pdf_reader.pages
for page in pdf_reader.pages:
  print(page.extract_text())
```

```python
file1=open(filename1,'rb')
pdf_reader=PyPDF2.PdfReader(file1)
pdf_reader.pages
print("pdf1")
for page in pdf_reader.pages:
  print(page.extract_text())
print("pdf2")
file2=open(filename2,'rb')
pdf_reader=PyPDF2.PdfReader(file2)
pdf_reader.pages
for page in pdf_reader.pages:
  print(page.extract_text())
```

```python
file=open('merged file 5.pdf','wb')
pdf_reader=PyPDF2.PdfWriter(file)
pdf_reader.pages
for page in pdf_reader.pages:
  print(page.extract_text())
```

#Split a PDF

```python
file1 = PdfReader(open(filename1, "rb"))
file2 = PdfReader(open(filename2, "rb"))

output = PdfWriter()

output.add_page(file1.pages[5])
output.add_page(file2.pages[3])


outputStream = open("merged file 2.pdf", "wb")
output.write(outputStream)
outputStream.close()
```
